## Numerical python

```python
sr = 2000.0
ts = 1.0/sr

t = np.arange(0., 1., ts)

angle_constant = 2.0 * np.pi

freq = 1.
x = 3.0 * np.sin(angle_constant * freq * t)

freq = 4.
x += 1. * np.sin(angle_constant * freq * t)

freq = 7.
x += 0.5 * np.sin(angle_constant * freq * t)

plt.plot(t, x)
plt.ylabel("Amplitude")
```

```python
X = np.fft.fft(x)
n = len(X)
ns = np.arange(n)

T = n/sr
freq = ns/T 
```

```python
plt.stem(freq, np.abs(X))
plt.xlim(0., 10.)
```
