```python
import numpy as np
import polars as pl
```

```python
n = 40

data = {
    "specie": np.repeat(["dragon", "turtle"], n/2),
    "temperature": np.round(np.random.normal(size=n, loc=90.0, scale=10.0))
}
```

```python
noise = np.array(
    [
        (
            np.random.normal(loc=0.0, scale=1.0)
            if specie == "dragon"
            else np.random.normal(loc=0.0, scale=0.5)
        )
        for specie in data["specie"]
    ]
)

data.update(
    {"girth": 10.0 + np.equal(data["specie"], "dragon") * -0.5 + data["temperature"] * 0.2 + noise}
)
```

```python
design_matrix = np.concatenate(
    [
        (np.equal(data["specie"], "dragon") * 1.)[:, None],
        data["temperature"][:, None],
        np.repeat(1., data["temperature"].shape[0])[:, None],
    ],
    axis=1,
)
```

```python
design_matrix[:5, :]
```

```python
coefs, *_ = np.linalg.lstsq(design_matrix, data["girth"], rcond=None)
coefs[0]
```
